### Motivating query and eval: Compound Expressions

```python
import numpy as np
rng = np.random.default_rng(42)
x = rng.random(1000000)
y = rng.random(1000000)
%timeit x + y
```

```python
%timeit np.fromiter((xi + yi for xi, yi in zip(x, y)),dtype=x.dtype, count=len(x))
```

```python
mask = (x > 0.5) & (y < 0.5)
```

```python
tmp1 = (x > 0.5)
tmp2 = (y < 0.5)
mask = tmp1 & tmp2
```

```python
import numexpr
mask_numexpr = numexpr.evaluate('(x > 0.5) & (y < 0.5)')
np.all(mask == mask_numexpr)
```

### pandas.eval for Efficient Operations

```python
import pandas as pd
nrows, ncols = 100000, 100
df1, df2, df3, df4 = (pd.DataFrame(rng.random((nrows, ncols)))
                      for i in range(4))
```

```python
#compute the sum of all four dataframes 
%timeit df1 + df2 + df3 + df4
```

```python
#compute the sum of all dataframes using pd.eval
%timeit pd.eval('df1 + df2 + df3 + df4')
```

```python
np.allclose(df1 + df2 + df3 + df4 + df5,
            pd.eval('df1 + df2 + df3 + df4 + df5'))
```
