```python
students = ['MERCY', 'JoHn', 'JaNE', 'dOe', 'NAncy']
[student.lower() for student in students]
```

```python
students = ['Eno', 'Carl', 'VICTOR', 'NIgel', None, 'Byron', 'Christine']
[student.capitalize() for student in students]
```

```python
import pandas as pd
names = pd.Series(students)
names
```
