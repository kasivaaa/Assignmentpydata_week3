```python
#import the libraries
import pandas as pd
import numpy as np
```

```python
#assign to a variable called iris
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris = pd.read_csv(url, header=None)
iris.head()
```

```python
#create columns for the dataset
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
iris.head()
```

```python
#is there any missing values
iris.isnull().sum()
```

```python
#  set the values of the rows 10 to 29 of the column 'petal_length' to NaN
iris.loc[10:29, 'petal_length'] = np.nan
iris.loc[10:29, 'petal_length']
```

```python
#substitute the NaN values to 1.0
iris['petal_length'] = iris['petal_length'].fillna(1.0)
iris.loc[10:29, 'petal_length']
```

```python
#delete the column class
iris = iris.drop(columns=['class'])
iris.head()
```

```python
#Set the first 3 rows as NaN
iris.iloc[0:3, :] = np.nan
iris.head()
```

```python
#Delete the rows that have NaN
iris = iris.dropna()
iris
```

```python
# Reset the index so it begins with 0 again
iris = iris.reset_index(drop=True)
iris.head()
```
