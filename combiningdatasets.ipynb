### Combining data sets

```python
import pandas as pd
import numpy as np
```

```python
def make_df(cols, ind):
	"""Quickly make a DataFrame"""
	data = {
		c: [str(c) + str(i) for i in ind]
		for c in cols
	}
	return pd.DataFrame(data, ind)

# example DataFrame
make_df("AB", range(3))
```

```python
class display(object):
	"""Display HTML representation of multiple objects"""
	template = """<div style="float: left; padding: 10px;">
	<p style='font-family:"Courier New", Courier, monospace'>{0}</p>{1}
	</div>"""

	def __init__(self, *args):
		self.args = args

	def _repr_html_(self):
		parts = []
		for a in self.args:
			obj = eval(a)
			html = obj._repr_html_() if hasattr(obj, "_repr_html_") else repr(obj)
			parts.append(self.template.format(a, html))
		return '\n'.join(parts)

	def __repr__(self):
		return '\n\n'.join(a + '\n' + repr(eval(a)) for a in self.args)
```

### Read Csv Function

```python
import pandas as pd 
pop = pd.read_csv('state-population.csv')
areas = pd.read_csv('state-areas.csv')
abbrevs = pd.read_csv('state-abbrevs.csv')

display(pop.head(), areas.head(), abbrevs.head())
```

###  Use how='outer' to make sure no data is thrown away due to mismatched labels

```python
# many-to-one merge that will give us the full state names within the population DataFrame

merged = pd.merge(pop,abbrevs, how = 'outer', left_on = 'state/region', right_on = 'abbreviation')
merged = merged.drop('abbreviation', axis = 1)
merged.head()
```

```python
# Displaying the final merged DataFrame with the areas DataFrame using a left join on the 'state' column
final_data =pd.merge(merged, areas, on = 'state', how = 'left')
final_data.head()
```

```python
data2010 = final_data.query("year == 2010 & ages == 'total'")
data2010.head()
```

```python
# Computing population density and displaying in order
# Reindexing the data2010 DataFrame by the 'state' column to facilitate density calculation
data2010.set_index('state', inplace=True)
density = data2010['population'] / data2010['area (sq. mi)']
```

```python
# sort values to see the start of the list
density.sort_values(ascending=False, inplace=True)
density.head()
```
