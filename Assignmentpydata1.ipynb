```python
#62.  Sum arrays of shape (1,3) and (3,1) using an iterator

A = np.arange(3).reshape((1, 3))
B = np.arange(3).reshape((3, 1))

iterator = np.nditer([A, B, None],
    flags=["external_loop"],
    op_flags=[
       ["readonly"],
       ["readonly"],
       ["writeonly", "allocate"]
    ]
)

for x, y, result in iterator:
    result[...] = x + y

print(iterator.operands[2])
```

```python
#63.  Create a class with a name attribute
class NamedArray(np.ndarray):

    def __new__(cls, values, name="No name"):
        obj = np.asarray(values).view(cls)
        obj.name = name
        return obj

    def __array_finalize__(self, obj):
        if obj is None:
            return

        self.name = getattr(obj, "name", "No name")


Z = NamedArray([1, 2, 3, 4], name="My Array")

print(Z)
print(Z.name)
```

```python
#78.  Distance from one point to several lines 
P0 = np.array([
    [0, 0],
    [1 ,1],
    [2, 0],
 ], dtype=float)

P1 = np.array([
     [4, 0],
     [1 ,5],
     [5, 3],
], dtype=float)

p = np.array([2, 2], dtype=float)

line_vectors = P1 - P0
point_vectors = p - P0

cross_products = (
     line_vectors[:, 0] * point_vectors[:, 1] - line_vectors[:, 1] * point_vectors[:, 0]

)
distances = np.abs(cross_products) / np.linalg.norm(line_vectors, axis=1)
print("Distances from point to lines:", distances)

```

```python
# 79. Distance from several points to several lines 
P0 = np.array([
    [0, 0],
    [1, 1]
], dtype=float)

P1 = np.array([
    [4, 0],
    [1, 5]
], dtype=float)

P = np.array([
    [2, 2],
    [3, 1],
    [0, 4]
], dtype=float)

line_vectors = P1 - P0

point_vectors = P[None, :, :] - P0[:, None, :]

cross_products = (
    line_vectors[:, None, 0] * point_vectors[:, :, 1]
    - line_vectors[:, None, 1] * point_vectors[:, :, 0]
)

distances = np.abs(cross_products) / np.linalg.norm(
    line_vectors,
    axis=1
)[:, None]

print(distances)
```

```python
# 80 Extract a centered subarray and pad where necessary 
def extract_subarray(array, centre, shape, fill=0):
    result = np.full(shape, fill, dtype=array.dtype)

    centre_row, centre_column = centre
    result_rows, result_columns = shape

    start_row = centre_row - result_rows // 2
    start_column = centre_column - result_columns // 2

    end_row = start_row + result_rows
    end_column = start_column + result_columns

    source_start_row = max(start_row, 0)
    source_start_column = max(start_column, 0)

    source_end_row = min(end_row, array.shape[0])
    source_end_column = min(end_column, array.shape[1])

    destination_start_row = source_start_row - start_row
    destination_start_column = source_start_column - start_column

    destination_end_row = (
        destination_start_row
        + source_end_row
        - source_start_row
    )

    destination_end_column = (
        destination_start_column
        + source_end_column
        - source_start_column
    )

    result[
        destination_start_row:destination_end_row,
        destination_start_column:destination_end_column
    ] = array[
        source_start_row:source_end_row,
        source_start_column:source_end_column
    ]

    return result


Z = np.arange(25).reshape(5, 5)

result = extract_subarray(
    Z,
    centre=(0, 0),
    shape=(3, 3),
    fill=-1
)

print(result)
```

```python
# 85. Create a symettric 2D array 
class SymmetricArray(np.ndarray):

    def __new__(cls, values):
        array = np.asarray(values)

        if array.ndim != 2 or array.shape[0] != array.shape[1]:
            raise ValueError("The array must be square.")

        symmetric = np.triu(array)
        symmetric = symmetric + symmetric.T - np.diag(
            np.diag(symmetric)
        )

        return symmetric.view(cls)

    def __setitem__(self, index, value):
        row, column = index

        np.ndarray.__setitem__(self, (row, column), value)
        np.ndarray.__setitem__(self, (column, row), value)


Z = SymmetricArray([
    [1, 2, 3],
    [0, 4, 5],
    [0, 0, 6]
])

print(Z)

Z[0, 2] = 20

print(Z)
print(Z[0, 2] == Z[2, 0])
```

```python
#90. Build a caretesian product
def cartesian_product(*arrays):
    grids = np.meshgrid(*arrays, indexing="ij")

    product = np.stack(
        grids,
        axis=-1
    ).reshape(-1, len(arrays))

    return product


A = np.array([1, 2])
B = np.array([10, 20, 30])
C = np.array([100, 200])

result = cartesian_product(A, B, C)

print(result)
```

```python
# 93. Find the rows from A containing Elements from B
A = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [2, 1, 8],
    [7, 8, 9],
    [5, 4, 0],
    [1, 7, 3],
    [6, 2, 4],
    [9, 8, 7]
])

B = np.array([
    [1, 2],
    [8, 9]
])

matches = (
    A[:, None, :, None]
    == B[None, :, None, :]
).any(axis=2).all(axis=2)

for row_number, values in enumerate(B):
    print(f"Rows containing {values}:")
    print(A[matches[:, row_number]])
```

```python
# 100 Bootsrap confidence
X = np.array([
    12, 15, 14, 10, 18,
    20, 13, 16, 17, 11
])

number_of_resamples = 10000

rng = np.random.default_rng(seed=42)

bootstrap_samples = rng.choice(
    X,
    size=(number_of_resamples, len(X)),
    replace=True
)

bootstrap_means = bootstrap_samples.mean(axis=1)

lower_limit, upper_limit = np.percentile(
    bootstrap_means,
    [2.5, 97.5]
)

print("Sample mean:", X.mean())
print("95% confidence interval:")
print(lower_limit, upper_limit)
```
